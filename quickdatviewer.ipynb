```python
import numpy as np
import pandas as pd
```

```python
data_file = '/home/kale-chen/Downloads/1_points_run2_HighInt_5min_coinc.dat'
```

```python
for chunk in pd.read_csv(data_file, sep = '\t', chunksize = 1000000, usecols = [2, 4, 7, 9], header = None):
    chunk.columns = ['TimeL', 'IDL', 'TimeR', 'IDR']
    chunk['TimeDiff'] = chunk['TimeL'] - chunk['TimeR'] / 2 # Calculate time of flight
    chunk['IDL'] = (chunk['IDL'] - 131072) % 3072 + 3072 # Convert IDL to lut form
    chunk['IDR'] = chunk['IDR'] % 3072 # Convert IDR to lut form    
    chunk = chunk[['TimeL', 'IDL', 'IDR', 'TimeDiff']] # Keep needed columns
    break
```
