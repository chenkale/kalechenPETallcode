```python
# Gif maker
import os
import imageio
```

```python
name = 'mcirc_pcut'
names = [f'{name}intertimes', f'{name}intertimescut1', f'{name}intertimescut2']
#names = [f'{name}proj', f'{name}projcut1', f'{name}projcut2']
img_dir = 'Sensitivity Normalization/plots'
img_paths = []
for name in names:
    img_paths.append(os.path.join(img_dir, f'{name}.png'))
img_paths.sort()
print(img_paths)
```

```python
outname = f'{name}intertimes.gif'
images = []
for img_path in img_paths:
    images.append(imageio.imread(img_path))
imageio.mimsave(os.path.join(img_dir, outname), images, fps=0.5, loop=0)
```
