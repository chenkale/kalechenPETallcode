```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from tqdm import tqdm
```

```python
df = pd.read_csv('longFEBDtoC&T.dat', sep = '\t', header = None, usecols = [2, 3, 4, 7, 8, 9])
df.columns = ['TimeL', 'EnergyL', 'IDL', 'TimeR', 'EnergyR', 'IDR']
df['IDL'] = df['IDL'] % 128
df['IDR'] = df['IDR'] % 128
df['EnergyL'] = df['EnergyL'] * 100
df['EnergyR'] = df['EnergyR'] * 100
data = np.asarray(df, dtype = np.int64)
del df
```

```python
minus_time = data[1:,[0, 3]] - data[:-1,[0, 3]]
counts = np.where(minus_time < 10000, 1, 0)
print(np.sum(counts))
```
