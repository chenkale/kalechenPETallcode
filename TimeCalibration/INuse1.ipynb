```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read calibration data into pd DataFrame
df = pd.DataFrame()
for chunk in pd.read_csv('C:/Users/burri/Downloads/MovingSource_LowInt_60s_Top_HWTrigOff_coinc.dat', usecols=[2, 4, 7, 9], delimiter="\t", header=None, chunksize=10000):
    # Name columns
    chunk.columns = ['TimeL', 'IDL', 'TimeR', 'IDR']
    # Convert Left channel to mod 3072
    #chunk['IDL'] = chunk['IDL']
    # Time difference column, left minus right lets positive time differences be on the right side
    chunk['TimeDiff'] = chunk['TimeL'] - chunk['TimeR']
    
    df = pd.concat([df, chunk], ignore_index=True)

# Convert to numpy array, faster
arr = df.to_numpy()
```

```python
temp = []
for row in arr:
    if row[1] not in temp:
        temp.append(row[1])
print(sorted(temp))
```
