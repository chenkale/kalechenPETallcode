```python
def y0(crystals, idl, idr):
    xL, yL, zL = crystals[idl + 96]
    xR, yR, zR = crystals[idr]
    # Get midpoint between two channels
    mX, mY = (xL+xR)/2, (yL+yR) / 2
    # If midpoint lies on x = 0, return 0
    if mX == 0:
        return mY
    # else need to calculate distance between x = 0 and midpoint along LOR
    else:
        # Scale factor sf = the ratio of the distance between mX and x=0 and the entire length of the LOR
        sf = - mX / (xR-xL)
        ylength = sf * (yR - yL)
        ypos = mY + ylength
        return ypos
```
