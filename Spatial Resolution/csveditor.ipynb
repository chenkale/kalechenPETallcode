```python
#
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

```

```python
df = pd.read_csv('Data2/resolutionsiterated.csv', sep = ',')
print(df)
```

```python
df['FWHMxC'] = np.sqrt(df['FWHMx']**2 - (1.4*0.68)**2).round(3)
df['FWHMyC'] = np.sqrt(df['FWHMy']**2 - (1.4*0.68)**2).round(3)
```

```python
block = 0
print(df[block*9:block*9+1])
print(df[block*9:(block+1)*9][['name', 'FWTMx', 'FWHMx', 'FWTMy', 'FWHMy', 'FWHMxC', 'FWHMyC']].to_string(index=False))
```

```python
df2 = pd.read_csv('Data2/resolutionsiterated.csv', sep = ',')
df2 = df2[180:]
```

```python
pos1723 = df2[df2['name'] == 1723]
pos0923 = df2[df2['name'] == 923]
pos1716 = df2[df2['name'] == 1716]
pos0916 = df2[df2['name'] == 916]
```

```python
xs = range(1, 31)
plt.plot(xs, pos1723['FWHMx'], label = '1723 FWHMx', color = 'blue')
plt.plot(xs, pos1723['FWHMy'], label = '1723 FWHMy', color = 'red')
plt.plot(xs, pos0923['FWHMx'], label = '923 FWHMx', color = 'blue', linestyle = '--')
plt.plot(xs, pos0923['FWHMy'], label = '923 FWHMy', color = 'red', linestyle = '--')
plt.plot(xs, pos1716['FWHMx'], label = '1716 FWHMx', color = 'blue', linestyle = 'dotted')
plt.plot(xs, pos1716['FWHMy'], label = '1716 FWHMy', color = 'red', linestyle = 'dotted')
plt.plot(xs, pos0916['FWHMx'], label = '916 FWHMx', color = 'blue', linestyle = '-.')
plt.plot(xs, pos0916['FWHMy'], label = '916 FWHMy', color = 'red', linestyle = '-.')
plt.legend()
plt.xlabel('Iteration')
plt.ylabel('FWHM (mm)')
plt.show()
```
